## **1. Business Understanding**

#### 1.1 Latar Belakang & Pernyataan Masalah (*Problem Statement & Stakeholders*)
Industri penerbangan komersial menerapkan strategi penetapan harga dinamis (*dynamic pricing*) di mana harga tiket dapat berfluktuasi secara tajam dalam hitungan hari atau jam. Fluktuasi ini dipengaruhi oleh kombinasi faktor yang rumit, seperti kelas kabin (Economy vs. Business), maskapai penerbangan, durasi penerbangan, jumlah pemberhentian (*transit*), serta waktu keberangkatan dan kedatangan.

Kondisi harga yang tidak menentu dan sulit ditebak ini menimbulkan permasalahan nyata bagi dua *stakeholders* utama:
1. **Calon Penumpang (Konsumen):** Mengalami kebingungan dan ketidakpastian dalam menentukan waktu terbaik serta pilihan tiket yang paling rasional dan ekonomis untuk anggaran mereka.
2. **Agen Perjalanan (*Travel Agents & Aggregators*):** Memerlukan sistem estimasi harga wajar untuk merancang strategi promosi yang kompetitif, memberikan rekomendasi tarif terbaik kepada pelanggan, serta memprediksi tren permintaan tiket.

Oleh karena itu, diperlukan suatu sistem analitik berbasis *Machine Learning* yang mampu mengestimasi harga tiket pesawat secara akurat berdasarkan fitur-fitur karakteristik penerbangan.


#### 1.2 Rumusan Masalah / Pertanyaan Bisnis (*Business Questions*)
Berdasarkan latar belakang masalah tersebut, pertanyaan analitis dan bisnis yang ingin dijawab dalam proyek ini adalah:
1. Faktor atau fitur penerbangan apa saja yang memiliki korelasi dan pengaruh paling signifikan terhadap harga tiket pesawat?
2. Seberapa besar disparitas harga antara penerbangan kelas *Economy* dan kelas *Business*?
3. Bagaimana pola variasi harga tiket berdasarkan masing-masing maskapai penerbangan dan rute perjalanan antarkota metropolitan?
4. Apakah penambahan jumlah transit (*stops*) dan durasi perjalanan berbanding lurus dengan peningkatan harga tiket?
5. Bagaimana merancang *pipeline data preprocessing* dan rekayasa fitur (*feature engineering*) yang optimal sehingga data mentah penerbangan siap diproses oleh model regresi machine learning?


### 1.3 Tujuan Proyek (SMART Goals)
Tujuan utama proyek ini adalah membangun model regresi yang memprediksi harga tiket (INR) dari karakteristik penerbangan dengan target $R^2 \geq 0,90$ dan MAPE $\leq 15\%$ pada data uji, yang diselesaikan dalam semester 2026/2027.

* **Measurable:** Kinerja model dievaluasi dengan target pencapaian $R^2 \geq 0,90$ dan MAPE $\leq 15\%$ pada data uji; metrik tambahan seperti MAE dan RMSE juga akan dilaporkan untuk semua model.
* **Achievable:** Pengerjaan dapat direalisasikan oleh tim beranggotakan 3 orang dengan menggunakan data tabular publik, algoritma standar dari *scikit-learn*, serta alur *preprocessing* yang umum.
* **Relevant:** Model dirancang untuk menjawab lima pertanyaan bisnis utama serta mengakomodasi kebutuhan agen perjalanan dan calon penumpang akan estimasi harga tiket pesawat yang wajar.

#### 1.4 Penentuan Metode Machine Learning
* **Metode yang Dipilih:** **Regresi (*Supervised Learning*)**.
* **Alasan Pemilihan:** Target prediksi proyek ini adalah `price` (harga tiket pesawat), yang merupakan variabel kontinu bernilai riil (angka/numerik). Sesuai materi kurikulum praktikum data science, tugas memprediksi nilai kuantitatif kontinu diselesaikan menggunakan pendekatan algoritma regresi (seperti *Linear Regression, Ridge/Lasso, Decision Tree Regressor, Random Forest Regressor*, dll.).


#### 1.5 Metrik Evaluasi Model (*Evaluation Metrics*)
Untuk mengukur keakuratan estimasi harga tiket pada tahap pemodelan nanti, metrik evaluasi yang ditetapkan adalah:
1. **Mean Absolute Error (MAE):** Mengukur rata-rata selisih absolut antara harga prediksi model dan harga riil dalam satuan mata uang (INR/Rupee), memberikan interpretasi intuitif rata-rata melesetnya prediksi harga.
2. **Root Mean Squared Error (RMSE):** Memberikan penalti bobot yang lebih besar terhadap kesalahan prediksi ekstrem (*large errors*).
3. **Mean Absolute Percentage Error (MAPE):** Mengukur rata-rata persentase deviasi kesalahan prediksi relatif terhadap harga sebenarnya.
4. **Coefficient of Determination ($R^2$ Score):** Mengukur seberapa besar proporsi variansi harga tiket yang berhasil dijelaskan oleh kombinasi fitur prediktor dalam model.

## **2. Data Gathering & Ingestion**


#### 2.1 Import Library
Mengimpor library Python yang esensial untuk pengolahan data, komputasi numerik, visualisasi data, serta preprocessing.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import warnings
from pathlib import Path
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from IPython.display import display

# Konfigurasi display dan visualisasi
warnings.filterwarnings('ignore')
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', lambda x: '%.2f' % x)
sns.set_theme(style='whitegrid')
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
```

#### 2.2 Load Dataset Mentah (`business.csv` & `economy.csv`)
Memuat dua dataset terpisah: data penerbangan kelas bisnis (`business.csv`) dan data penerbangan kelas ekonomi (`economy.csv`).


```python
# Memeriksa keberadaan file dataset mentah
business_path = Path("business.csv")
economy_path = Path("economy.csv")

if not business_path.exists() or not economy_path.exists():
    raise FileNotFoundError("File business.csv dan economy.csv tidak ditemukan di direktori kerja!")

df_business = pd.read_csv(business_path)
df_economy = pd.read_csv(economy_path)

print(f"Dimensi Data Business : {df_business.shape[0]:,} baris, {df_business.shape[1]} kolom")
print(f"Dimensi Data Economy  : {df_economy.shape[0]:,} baris, {df_economy.shape[1]} kolom")

print("\nContoh 3 baris pertama data Business:")
display(df_business.head(3))

print("Contoh 3 baris pertama data Economy:")
display(df_economy.head(3))
```

#### 2.3 Penyatuan Skema Data & Vertical Concatenation
Kedua dataset memiliki nama kolom yang identik, namun belum mencantumkan pembeda kelas layanan secara eksplisit. Oleh karena itu, kita menambahkan kolom `class` bernilai `'Business'` pada data bisnis dan `'Economy'` pada data ekonomi, lalu menggabungkan keduanya secara vertikal (*row-wise concatenation*).


```python
# Menambahkan kolom penanda kelas layanan
df_business = df_business.copy()
df_economy = df_economy.copy()

df_business['class'] = 'Business'
df_economy['class'] = 'Economy'

# Menggabungkan kedua dataframe menjadi satu dataset gabungan
df = pd.concat([df_business, df_economy], ignore_index=True)

print(f"Total baris dataset gabungan : {df.shape[0]:,} baris")
print(f"Total kolom dataset gabungan : {df.shape[1]} kolom")
print(f"Daftar kolom: {df.columns.tolist()}")

display(df.sample(5, random_state=42))
```

## **3. Exploratory Data Analysis (EDA)**
Tahap EDA bertujuan untuk menganalisis karakteristik statistik data, memetakan bentuk sebaran variabel target `price`, mengevaluasi korelasi dan pengaruh masing-masing fitur prediktor terhadap harga tiket, serta mengidentifikasi pola anomali/outlier sebelum penanganan data.


#### 3.1 Statistik Deskriptif Awal
Kita membuat salinan dataset untuk keperluan analisis eksploratif dan mengonversi format kolom harga sementara menjadi numerik agar dapat dihitung statistik deskriptifnya.


```python
# Membuat dataframe salinan untuk eksplorasi
df_eda = df.copy()

# Konversi sementara kolom price menjadi numerik untuk eksplorasi statistik
df_eda['price_num'] = pd.to_numeric(df_eda['price'].astype(str).str.replace(',', '', regex=False), errors='coerce')

print("=== Statistik Deskriptif Kolom Numerik ===")
display(df_eda[['num_code', 'price_num']].describe().T)

print("\n=== Statistik Deskriptif Kolom Kategorikal ===")
display(df_eda.describe(include=['object']).T)
```

*Insight:*
1. **Sebaran Harga:** Harga tiket pesawat (`price_num`) memiliki rentang variasi yang sangat ekstrem, mulai dari nilai minimum **INR 1.105** hingga nilai maksimum mencapai **INR 123.071**, dengan rata-rata **INR 20.883,72** dan standar deviasi sebesar **INR 22.697,79**.
2. **Kemiringan Data (*Skewness*):** Nilai mean (INR 20.883) berada jauh di atas nilai median (INR 7.425, persentil ke-50). Hal ini mengindikasikan bahwa distribusi harga tiket sangat menceng ke kanan (*heavily right-skewed*), yang dipicu oleh perbedaan harga masif antara kelas Business dan Economy.
3. **Karakteristik Kategorikal:** Terdapat 8 maskapai penerbangan unik (dengan frekuensi terbanyak adalah Vistara), serta 6 kota rute metropolitan utama dengan Delhi dan Mumbai sebagai simpul transportasi terpadat.


#### 3.2 Distribusi Variabel Target (Harga Tiket / Price)
Visualisasi sebaran variabel target `price` secara keseluruhan menggunakan kurva histogram dan Kernel Density Estimation (KDE), baik pada skala asli maupun skala logaritmik.


```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Plot 1: Distribusi Harga Asli
sns.histplot(df_eda['price_num'], bins=60, kde=True, color='#2b5c8f', ax=axes[0])
axes[0].set_title('Distribusi Harga Tiket Pesawat (Skala Asli)', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Harga Tiket (INR)', fontsize=11)
axes[0].set_ylabel('Frekuensi', fontsize=11)

# Plot 2: Distribusi Log-Transformed Harga
sns.histplot(np.log1p(df_eda['price_num']), bins=60, kde=True, color='#16a085', ax=axes[1])
axes[1].set_title('Distribusi Harga Tiket (Log-Transformed)', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Log(Harga + 1)', fontsize=11)
axes[1].set_ylabel('Frekuensi', fontsize=11)

plt.tight_layout()
plt.show()
```

*Insight:*
1. **Pola Bimodal:** Distribusi harga tiket secara keseluruhan memperlihatkan pola *bimodal* (dua puncak frekuensi yang terpisah). Puncak pertama berada pada rentang harga murah (INR 2.000 – INR 10.000) dan puncak kedua berada pada rentang harga tinggi (INR 40.000 – INR 70.000).
2. Pola dua puncak ini secara alami mencerminkan dua jenis produk layanan yang berbeda drastis: kelas *Economy* dan kelas *Business*.
3. Transformasi logaritmik memperjelas pemisahan kedua klaster harga tersebut, yang mengindikasikan bahwa fitur kelas layanan (`class`) akan memegang peranan krusial sebagai fitur prediktor utama dalam regresi.


#### 3.3 Analisis Perbandingan Harga Berdasarkan Kelas (Business vs. Economy)
Menganalisis disparitas harga tiket penerbangan antara kelas *Business* dan kelas *Economy* menggunakan visualisasi *boxplot* dan *density plot*.


```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Boxplot Perbandingan Harga per Kelas
sns.boxplot(data=df_eda, x='class', y='price_num', palette=['#e74c3c', '#3498db'], ax=axes[0])
axes[0].set_title('Distribusi Boxplot Harga: Business vs Economy', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Kelas Penerbangan', fontsize=11)
axes[0].set_ylabel('Harga Tiket (INR)', fontsize=11)

# KDE Plot Terpisah per Kelas
sns.kdeplot(data=df_eda, x='price_num', hue='class', common_norm=False, fill=True, palette=['#e74c3c', '#3498db'], ax=axes[1])
axes[1].set_title('Density Plot Sebaran Harga per Kelas', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Harga Tiket (INR)', fontsize=11)
axes[1].set_ylabel('Densitas', fontsize=11)

plt.tight_layout()
plt.show()

# Ringkasan Statistik Harga per Kelas
class_summary = df_eda.groupby('class')['price_num'].agg(
    Jumlah='count',
    Min='min',
    Q25=lambda x: x.quantile(0.25),
    Median='median',
    Mean='mean',
    Q75=lambda x: x.quantile(0.75),
    Max='max',
    Std='std'
)
print("Statistik Komparasi Harga Berdasarkan Kelas:")
display(class_summary)
```

*Insight:*
1. **Disparitas Harga Sangat Masif:** Terdapat jurang harga yang sangat nyata antara kedua kelas. Tiket kelas Economy memiliki rata-rata harga **INR 6.572,35** (median INR 5.454), sedangkan kelas Business memiliki rata-rata harga **INR 52.540,09** (median **INR 53.164**) — hampir **8 kali lipat lebih mahal**.
2. **Karakteristik Outlier Kontekstual:** Pada boxplot kelas Economy terlihat banyak titik observasi di atas batas atas whisker (harga tiket ekonomi > INR 15.000). Namun, setelah ditelusuri, titik-titik tersebut **bukanlah data error/palsu**, melainkan tiket penerbangan ekonomi jarak jauh, mendekati hari keberangkatan (*urgent/last-minute*), atau dengan multi-transit.
3. **Implikasi Cleaning:** Kita tidak boleh sembarangan menghapus data bernilai tinggi sebagai outlier global, karena nilai tinggi tersebut merupakan data valid dari kelas Business dan tiket urgent.


#### 3.4 Analisis Pengaruh Maskapai Penerbangan (Airline) terhadap Harga
Menganalisis rata-rata harga tiket dan keterlibatan maskapai dalam melayani penerbangan kelas Economy maupun Business.


```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# Plot 1: Rata-rata Harga Keseluruhan per Maskapai
airline_avg_order = df_eda.groupby('airline')['price_num'].mean().sort_values(ascending=False).index

sns.barplot(data=df_eda, x='airline', y='price_num', order=airline_avg_order, palette='crest', ax=axes[0])
axes[0].set_title('Rata-rata Harga Tiket per Maskapai Penerbangan', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Maskapai', fontsize=11)
axes[0].set_ylabel('Rata-rata Harga (INR)', fontsize=11)
axes[0].tick_params(axis='x', rotation=45)

# Plot 2: Rata-rata Harga Maskapai Terpilah Berdasarkan Kelas
sns.barplot(data=df_eda, x='airline', y='price_num', hue='class', palette=['#e74c3c', '#3498db'], ax=axes[1])
axes[1].set_title('Rata-rata Harga Maskapai Berdasarkan Kelas Layanan', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Maskapai', fontsize=11)
axes[1].set_ylabel('Rata-rata Harga (INR)', fontsize=11)
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# Tabel Kontingensi Frekuensi Maskapai vs Kelas
print("Frekuensi Penerbangan Tiap Maskapai Berdasarkan Kelas:")
display(pd.crosstab(df_eda['airline'], df_eda['class'], margins=True))
```

*Insight:*
1. **Maskapai Full-Service vs Low-Cost:** Maskapai **Vistara** dan **Air India** memiliki rata-rata harga tertinggi secara keseluruhan karena keduanya merupakan satu-satunya maskapai dalam dataset yang menyediakan layanan kabin kelas **Business**.

#### 4.4 Feature Engineering (Pembuatan Fitur Baru)
Membentuk fitur-fitur baru yang lebih informatif dan terstandarisasi untuk meningkatkan performa algoritma regresi:
1. `duration_hours`: Mengekstrak durasi string (`'02h 15m'`) menjadi nilai numerik kontinu jam.
2. `stops`: Membersihkan whitespace liar dan standarisasi jumlah transit (`'zero'`, `'one'`, `'two_or_more'`).
3. `departure_time` & `arrival_time`: Melakukan *binning* jam keberangkatan dan kedatangan ke dalam 6 interval waktu hari.
4. `day_of_week` & `is_weekend`: Mengekstrak nama hari dan indikator akhir pekan dari tanggal penerbangan.


```python
# 1. Konversi time_taken ke duration_hours
def parse_duration(val):
    if pd.isna(val):
        return np.nan
    val = str(val).strip()
    h = 0.0
    m = 0.0
    mh = re.search(r'(\d+)\s*h', val)
    mm = re.search(r'(\d+)\s*m', val)
    if mh: h = float(mh.group(1))
    if mm: m = float(mm.group(1))
    return round(h + (m / 60.0), 2)

print("Transformasi 'time_taken' -> 'duration_hours':")
print(f"SEBELUM : {df['time_taken'].head(3).tolist()}")
df['duration_hours'] = df['time_taken'].apply(parse_duration)
print(f"SESUDAH : {df['duration_hours'].head(3).tolist()}")

# 2. Standarisasi kolom stop -> stops
def standardize_stops(val):
    val = str(val).strip().lower()
    if 'non-stop' in val or '0' in val:
        return 'zero'
    elif '1-stop' in val:
        return 'one'
    elif '2+' in val or '2-stop' in val:
        return 'two_or_more'
    return 'other'

print("\nTransformasi 'stop' -> 'stops':")
print(f"SEBELUM (Contoh mentah kotor):\n{df['stop'].value_counts().head(3)}")
df['stops'] = df['stop'].apply(standardize_stops)
print(f"SESUDAH (Kategori bersih):\n{df['stops'].value_counts()}")

# 3. Binning waktu keberangkatan dan kedatangan
def get_time_period(val):
    try:
        hour = int(str(val).strip().split(':')[0])
        if 4 <= hour < 8:
            return 'Early_Morning'
        elif 8 <= hour < 12:
            return 'Morning'
        elif 12 <= hour < 16:
            return 'Afternoon'
        elif 16 <= hour < 20:
            return 'Evening'
        elif 20 <= hour < 24:
            return 'Night'
        else:
            return 'Late_Night'
    except:
        return 'Unknown'

df['departure_time'] = df['dep_time'].apply(get_time_period)
df['arrival_time'] = df['arr_time'].apply(get_time_period)

# 4. Ekstraksi fitur kalender
df['flight_date'] = pd.to_datetime(df['date'], format='%d-%m-%Y', errors='coerce')
df['day_of_week'] = df['flight_date'].dt.day_name()
df['is_weekend'] = df['flight_date'].dt.dayofweek.isin([5, 6]).astype(int)

print("\nContoh Fitur Baru Hasil Feature Engineering:")
display(df[['duration_hours', 'stops', 'departure_time', 'arrival_time', 'day_of_week', 'is_weekend']].head(4))
```

*Insight:*
Tahap *Feature Engineering* sukses mentransformasikan data kotor menjadi atribut analitik yang terstruktur:
1. `duration_hours`: Menghasilkan fitur kontinu numerik yang akurat (rentang 0,83 jam hingga 49,83 jam dengan rata-rata 12,22 jam).
2. `stops`: Berhasil membersihkan karakter escape `\n\t` liar menjadi tiga kategori standar (`one`: 250.927 baris; `zero`: 36.044 baris; `two_or_more`: 13.288 baris).
3. `departure_time` & `arrival_time`: Mengubah jam menit acak menjadi 6 segmen perilaku perjalanan yang bermakna bagi konsumen dan model.
4. `day_of_week` & `is_weekend`: Menyediakan variabel temporal untuk menangkap pola lonjakan harga pada akhir pekan (*weekend price surge*).


#### 4.5 Analisis & Penanganan Outlier
Menganalisis pencilan (*outliers*) pada fitur `price` dan `duration_hours` dengan metode Interquartile Range (IQR) serta pembahasan pertimbangan konteks domain bisnis.


```python
print("=== Analisis Outlier IQR Berdasarkan Kelas Penerbangan ===")
for cls in ['Economy', 'Business']:
    sub_data = df[df['class'] == cls]['price']
    q1 = sub_data.quantile(0.25)
    q3 = sub_data.quantile(0.75)
    iqr = q3 - q1
    lower_lim = q1 - 1.5 * iqr
    upper_lim = q3 + 1.5 * iqr
    outliers = sub_data[(sub_data < lower_lim) | (sub_data > upper_lim)]
    print(f"Kelas {cls:8s}: Batas Bawah = INR {lower_lim:,.0f} | Batas Atas = INR {upper_lim:,.0f} | Outlier = {len(outliers):,} baris ({len(outliers)/len(sub_data)*100:.2f}%)")

# Visualisasi Boxplot Distribusi
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(data=df, x='class', y='price', palette=['#e74c3c', '#3498db'], ax=axes[0])
axes[0].set_title('Sebaran Outlier Harga Berdasarkan Kelas', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Kelas', fontsize=11)
axes[0].set_ylabel('Harga (INR)', fontsize=11)

sns.boxplot(data=df, x='stops', y='duration_hours', palette='viridis', ax=axes[1])
axes[1].set_title('Sebaran Durasi Penerbangan Berdasarkan Transit', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Jumlah Transit', fontsize=11)
axes[1].set_ylabel('Durasi (Jam)', fontsize=11)

plt.tight_layout()
plt.show()
```

*Insight:*
1. **Keputusan Kontekstual Outlier:** 
   Secara statistik IQR murni, terdapat titik-titik data di luar pagar batas atas harga (misal tiket ekonomi > INR 15.000 atau bisnis > INR 85.000). Namun, **data tersebut TIDAK DIHAPUS**.
2. **Alasan Bisnis:** 
   Harga tiket penerbangan yang tinggi merupakan transaksi riil yang sah di industri penerbangan, yang mencerminkan pemesanan mendesak pada hari-H (*last-minute bookings*), periode liburan puncak (*peak season*), atau tiket dengan fleksibilitas pembatalan penuh. Menghapus data tersebut secara artifisial akan membiaskan pemahaman model dan menyebabkan model meremehkan (*underestimate*) harga tiket pada kondisi permintaan tinggi.
3. **Penerbangan Berdurasi Panjang:**
   Observasi durasi di atas 30 jam merupakan rute dengan 2 kali transit atau waktu singgah (*layover*) yang panjang di bandara penghubung, yang sepenuhnya valid dan wajar secara operasional.

Mempertahankannya akan menimbulkan *multikolinearitas* dan redundansi data.
   - `flight_date`: Dibuang karena tipe tanggal mentah (*datetime*) tidak dapat diproses secara langsung oleh algoritma regresi numerik.

#### 4.7 Identifikasi & Encoding Fitur Kategorikal
Mengubah variabel kategorikal menjadi representasi numerik:
1. **Ordinal Encoding:** Diterapkan pada variabel berjenjang/bertingkat (`class` dan `stops`).
2. **One-Hot Encoding:** Diterapkan pada variabel nominal tanpa hierarki (`airline`, `from`, `to`, `departure_time`, `arrival_time`, `day_of_week`) dengan `drop_first=True`.


```python
# 1. Ordinal Encoding
# Economy (0) < Business (1)
# zero (0) < one (1) < two_or_more (2)
class_map = {'Economy': 0, 'Business': 1}
stops_map = {'zero': 0, 'one': 1, 'two_or_more': 2}

df_selected['class_encoded'] = df_selected['class'].map(class_map)
df_selected['stops_encoded'] = df_selected['stops'].map(stops_map)

# 2. One-Hot Encoding untuk variabel nominal
nominal_cols = ['airline', 'from', 'to', 'departure_time', 'arrival_time', 'day_of_week']

df_encoded = pd.get_dummies(df_selected, columns=nominal_cols, drop_first=True, dtype=int)

# Menghapus kolom kategorikal teks asli yang sudah diwakili oleh kolom ordinal
df_encoded = df_encoded.drop(columns=['class', 'stops'])

print(f"Dimensi Data SEBELUM Encoding : {df_selected.shape}")
print(f"Dimensi Data SESUDAH Encoding : {df_encoded.shape}")

print("\nDaftar Fitur Hasil Encoding (10 kolom pertama):")
print(df_encoded.columns[:10].tolist())
display(df_encoded.head(3))
```

*Insight:*
- **Ordinal Encoding** sangat cocok untuk `class` dan `stops` karena memiliki tingkatan nilai yang logis (Business lebih tinggi dari Economy; transit 2 lebih banyak dari 1). Ini menjaga informasi hierarki matematis tanpa menambah dimensi kolom.
- **One-Hot Encoding** digunakan pada fitur nominal seperti maskapai, rute asal, rute tujuan, serta interval waktu hari.
- Parameter `drop_first=True` diterapkan untuk membuang satu kategori basis (*reference category*) pada tiap fitur guna mencegah **Dummy Variable Trap** (kondisi multikolinearitas sempurna), yang sangat krusial agar matriks fitur stabil dan invertible pada model regresi linear.


#### 4.8 Feature Scaling
Menerapkan teknik **MinMaxScaler** pada fitur numerik kontinu (`duration_hours`) agar rentang nilainya seragam dalam skala $[0, 1]$, selaras dengan fitur-fitur biner $(0/1)$ hasil One-Hot Encoding.


```python
scaler = MinMaxScaler()

print("Kondisi SEBELUM Scaling:")
print(f"Min 'duration_hours'  : {df_encoded['duration_hours'].min():.2f}")
print(f"Max 'duration_hours'  : {df_encoded['duration_hours'].max():.2f}")
print(f"Mean 'duration_hours' : {df_encoded['duration_hours'].mean():.2f}")

# Menerapkan scaling
df_encoded['duration_hours_scaled'] = scaler.fit_transform(df_encoded[['duration_hours']])

print("\nKondisi SESUDAH Scaling:")
print(f"Min 'duration_hours_scaled'  : {df_encoded['duration_hours_scaled'].min():.2f}")
print(f"Max 'duration_hours_scaled'  : {df_encoded['duration_hours_scaled'].max():.2f}")
print(f"Mean 'duration_hours_scaled' : {df_encoded['duration_hours_scaled'].mean():.2f}")
```

*Insight:*
- **Mengapa MinMaxScaler:** Karena sebagian besar fitur dalam dataset kini merupakan variabel biner $[0, 1]$ hasil dari One-Hot Encoding dan indikator boolean. Dengan mentransformasi `duration_hours` ke rentang $[0, 1]$, seluruh fitur prediktor berada pada rentang skala yang setara.
